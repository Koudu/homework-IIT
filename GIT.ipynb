#Создание ветки 

```python
git branch report # создание новой ветки
git checkout report # переключение на новую ветку
git add . # индексация файлов для коммита
git commit -m "v1" # фиксация всех изменений
```

изображение.png

#Слияние файла в основную ветку

```python
git checkout HEAD
git merge report
```

#Публицакия файла в GitHub

```python
git remote add origin https://github.com/Koudu/homework-IIT.git
git branch -M main
git push -u origin main
```
